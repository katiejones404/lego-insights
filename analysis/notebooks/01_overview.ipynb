```python
import sqlite3
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DB_PATH = Path("../../data/lego.db")
conn = sqlite3.connect(DB_PATH)
```

## Sets per year and average complexity

```python
sets_by_year = pd.read_sql_query("SELECT * FROM agg_sets_by_year", conn)
sets_by_year.describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(sets_by_year["year"], sets_by_year["num_sets"])
axes[0].set_title("Sets released per year")

axes[1].plot(sets_by_year["year"], sets_by_year["avg_parts"])
axes[1].set_title("Average parts per set")

fig.tight_layout()
```
